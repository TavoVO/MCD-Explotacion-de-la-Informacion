```python
import matplotlib.pyplot as plt

# 1. Filtramos para obtener solo las reservaciones exitosas ('Check-Out')
df_checkout_hotel = df[df["reservation_status"] == "Check-Out"]

# 2. Agrupamos por semana del año y por tipo de hotel
checkouts_por_hotel_semana = df_checkout_hotel.groupby(["arrival_date_week_number", "hotel"]).size().unstack(fill_value=0)

# 3. Graficamos ambas líneas en el mismo lienzo
plt.figure(figsize=(14, 6))
plt.plot(
    checkouts_por_hotel_semana.index,
    checkouts_por_hotel_semana["City Hotel"],
    color='royalblue',
    linewidth=2,
    marker='o',
    markersize=4,
    label='City Hotel'
)
plt.plot(
    checkouts_por_hotel_semana.index,
    checkouts_por_hotel_semana["Resort Hotel"],
    color='darkorange',
    linewidth=2,
    marker='s',
    markersize=4,
    label='Resort Hotel'
)

plt.title("Distribución de Reservaciones 'Check-Out' por Semana del Año y Tipo de Hotel", fontsize=14)
plt.xlabel("Número de Semana (arrival_date_week_number)", fontsize=12)
plt.ylabel("Cantidad Acumulada de Reservaciones", fontsize=12)
plt.xlim(1, 53)
plt.xticks(range(1, 54, 2))
plt.legend(title="Tipo de Hotel", fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Calculamos las métricas por tipo de depósito: total, canceladas y tasa de cancelación
resumen_deposito = df.groupby("deposit_type").agg(
    reservas_totales=("is_canceled", "count"),
    reservas_canceladas=("is_canceled", "sum"),
    tasa_cancelacion=("is_canceled", lambda x: x.mean() * 100)
).reset_index()

# 2. Mostramos los datos detallados en la tabla
print("Análisis de cancelaciones por tipo de depósito:")
display(resumen_deposito)

# 3. Graficamos los resultados
plt.figure(figsize=(8, 5))
sns.barplot(
    x="deposit_type",
    y="tasa_cancelacion",
    data=resumen_deposito,
    palette="viridis",
    hue="deposit_type",
    legend=False
)

plt.title("Tasa de Cancelación según el Tipo de Depósito", fontsize=14)
plt.xlabel("Tipo de Depósito (deposit_type)", fontsize=12)
plt.ylabel("Tasa de Cancelación (%)", fontsize=12)
plt.ylim(0, 105)

# Añadimos las etiquetas de porcentaje sobre cada barra
for index, row in resumen_deposito.iterrows():
    plt.text(
        index,
        row["tasa_cancelacion"] + 2,
        f"{row['tasa_cancelacion']:.1f}%",
        ha='center',
        va='bottom',
        fontsize=11,
        weight='bold'
    )

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## Creación de Variables

```python
# Calculamos el total de noches combinando días de fin de semana y días de semana
df_clean["total_nights"] = (
    df_clean["stays_in_week_nights"]
    + df_clean["stays_in_weekend_nights"]
)

# Calculamos el total de huéspedes sumando adultos, niños y bebés hospedados
df_clean["total_guests"] = (
    df_clean["adults"]
    + df_clean["children"]
    + df_clean["babies"]
)
```

```python
# Variable: Viaja con niños o bebés (has_children)
df_clean["has_children"] = ((df_clean["children"] > 0) | (df_clean["babies"] > 0)).astype(int)

# Variable: Anticipación en meses (lead_time_months)
df_clean["lead_time_months"] = round(df_clean["lead_time"] / 30, 1)

# Variable: Ingresos totales estimados por reserva (revenue)
df_clean["revenue"] = df_clean["adr"] * df_clean["total_nights"]

# Variable: Habitación cambiada (room_type_changed)
df_clean["room_type_changed"] = (df_clean["reserved_room_type"] != df_clean["assigned_room_type"]).astype(int)

# Mostramos una muestra de las nuevas variables para comprobar que todo se creó correctamente
columnas_nuevas = ["children", "babies", "has_children", "lead_time", "lead_time_months", "adr", "total_nights", "revenue", "reserved_room_type", "assigned_room_type", "room_type_changed"]
display(df_clean[columnas_nuevas].sample(10))
```
